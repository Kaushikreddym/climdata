# NEX-GDDP-CMIP6 — climdata Example

**NEX-GDDP-CMIP6** is NASA's globally downscaled, bias-corrected CMIP6 product at 0.25° (~25 km) daily resolution.

This example runs the full workflow for a single point: extract → impute gaps → compute a climate extreme index → write NetCDF.

## Extract, Impute, and Compute an Index

```python
from climdata import ClimData
import pandas as pd
import xarray as xr

import logging

logging.basicConfig(
    level=logging.INFO,
    format="%(levelname)s | %(message)s",
    force=True,
)
```

```python
overrides = [
    "dataset=nexgddp",  # Choose the NEX-GDDP dataset for extraction
    "lat=52",
    "lon=13",
    "time_range.start_date=1989-01-01",  # Start date for data extraction
    "time_range.end_date=1989-12-31",    # End date for data extraction
    "variables=[tasmin]",       # Variables to extract: min/max temp and precipitation
    "data_dir=./data",    # Local directory to store raw/intermediate files
    "source_id=GFDL-ESM4",
    "index=tn10p",  # Climate extreme index to calculate
    "impute=BRITS"
]

# -----------------------------
# Step 1: Define the workflow sequence
# -----------------------------
seq = ["extract", "impute", "calc_index", "to_nc"]

# -----------------------------
# Step 2: Initialize the ClimData extractor
# -----------------------------
extractor = ClimData(overrides=overrides)

# -----------------------------
# Step 3: Run the Multi-Step workflow
# -----------------------------
result = extractor.run_workflow(
    actions=seq,
)
```
